```python
# ============================================
# STANDARDIZE COLUMN NAMES
# ============================================

def clean_column_names(df):
    df = df.copy()
    
    df.columns = (
        df.columns
        .str.strip()
        .str.lower()
        .str.replace(" ", "_")
        .str.replace("-", "_")
    )
    
    return df


candidates_clean = clean_column_names(candidates_clean)
results_clean = clean_column_names(results_clean)
winners_clean = clean_column_names(winners_clean)

print("Column names standardized.")
```

```python
print("Candidates:")
print(candidates_clean.columns.tolist())

print("\nResults:")
print(results_clean.columns.tolist())

print("\nWinners:")
print(winners_clean.columns.tolist())
```

```python
# ============================================
# REMOVE COMPLETELY EMPTY ROWS
# ============================================

candidates_clean = candidates_clean.dropna(how="all").copy()
results_clean = results_clean.dropna(how="all").copy()
winners_clean = winners_clean.dropna(how="all").copy()

print("Empty rows removed.")
```

```python
# ============================================
# REMOVE EXACT DUPLICATES
# ============================================

before_candidates = len(candidates_clean)
before_results = len(results_clean)
before_winners = len(winners_clean)

candidates_clean = candidates_clean.drop_duplicates().copy()
results_clean = results_clean.drop_duplicates().copy()
winners_clean = winners_clean.drop_duplicates().copy()

print("Candidates duplicates removed:", before_candidates - len(candidates_clean))
print("Results duplicates removed   :", before_results - len(results_clean))
print("Winners duplicates removed   :", before_winners - len(winners_clean))
```

```python
# ============================================
# CLEAN TEXT VALUES
# ============================================

def clean_text_columns(df):
    df = df.copy()
    
    text_columns = df.select_dtypes(
        include=["object", "string"]
    ).columns
    
    for column in text_columns:
        df[column] = (
            df[column]
            .astype("string")
            .str.strip()
        )
    
    return df


candidates_clean = clean_text_columns(candidates_clean)
results_clean = clean_text_columns(results_clean)
winners_clean = clean_text_columns(winners_clean)

print("Text values cleaned.")
```

```python
# ============================================
# STANDARDIZE EMPTY VALUES
# ============================================

def standardize_missing_values(df):
    df = df.copy()
    
    df = df.replace({
        "": np.nan,
        "NA": np.nan,
        "N/A": np.nan,
        "null": np.nan,
        "NULL": np.nan,
        "-": np.nan
    })
    
    return df


candidates_clean = standardize_missing_values(candidates_clean)
results_clean = standardize_missing_values(results_clean)
winners_clean = standardize_missing_values(winners_clean)

print("Missing-value representation standardized.")
```

```python
# ============================================
# CLEANING VALIDATION
# ============================================

datasets_clean = {
    "Candidates": candidates_clean,
    "Results": results_clean,
    "Winners": winners_clean
}

for name, df in datasets_clean.items():
    print(f"\n{'=' * 50}")
    print(name)
    print(f"{'=' * 50}")
    print("Rows:", len(df))
    print("Columns:", len(df.columns))
    print("Duplicates:", df.duplicated().sum())
    print("Missing values:", df.isnull().sum().sum())
```

```python
# ============================================
# MODULE 6: IDENTIFY ELECTION FIELDS
# ============================================

def find_columns(df):
    result = {}
    
    for column in df.columns:
        name = column.lower()
        
        if any(x in name for x in ["state", "st_name", "state_name"]):
            result.setdefault("state", []).append(column)
            
        if any(x in name for x in ["constituency", "pc_name", "pc_no"]):
            result.setdefault("constituency", []).append(column)
            
        if any(x in name for x in ["candidate", "cand_name"]):
            result.setdefault("candidate", []).append(column)
            
        if any(x in name for x in ["party", "partyname", "party_name"]):
            result.setdefault("party", []).append(column)
            
        if any(x in name for x in ["vote", "votes", "totvotpoll"]):
            result.setdefault("votes", []).append(column)
            
        if any(x in name for x in ["elector", "electors"]):
            result.setdefault("electors", []).append(column)
            
        if any(x in name for x in ["margin", "majority"]):
            result.setdefault("margin", []).append(column)
    
    return result


for name, df in datasets_clean.items():
    print(f"\n{'=' * 60}")
    print(name)
    print(f"{'=' * 60}")
    
    fields = find_columns(df)
    
    for field, columns in fields.items():
        print(f"{field:15} → {columns}")
```

```python
# ============================================
# MODULE 27: WINNING MARGIN
# ============================================

def calculate_winning_margin(group):

    sorted_votes = (
        group["total_votes"]
        .sort_values(ascending=False)
        .values
    )

    if len(sorted_votes) >= 2:
        return sorted_votes[0] - sorted_votes[1]

    return np.nan


margin_df = (
    vote_analysis
    .groupby([state_col, constituency_col])
    .apply(calculate_winning_margin)
    .reset_index(name="winning_margin")
)

display(margin_df.head(10))
```

```python
# ============================================
# MODULE 30.4: CLEAN HISTORICAL DATA
# ============================================

historical_lok_sabha = lok_sabha.copy()

# Remove duplicate rows
historical_lok_sabha = (
    historical_lok_sabha
    .drop_duplicates()
    .reset_index(drop=True)
)

# Convert numeric columns
historical_lok_sabha["year"] = pd.to_numeric(
    historical_lok_sabha["year"],
    errors="coerce"
)

historical_lok_sabha["totvotpoll"] = pd.to_numeric(
    historical_lok_sabha["totvotpoll"],
    errors="coerce"
)

historical_lok_sabha["electors"] = pd.to_numeric(
    historical_lok_sabha["electors"],
    errors="coerce"
)

print("Historical data cleaned.")
print("Rows:", len(historical_lok_sabha))
```
